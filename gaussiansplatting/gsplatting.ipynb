```python
import numpy as np
import torch
import matplotlib.pyplot as plt
import math
```

```python
mu = torch.randn(3,1)
def rotation_matrix(phi_deg):
    phi = phi_deg * math.pi / 180 
    c, s = torch.cos(phi), torch.sin(phi)
    z, o = torch.zeros_like(c), torch.ones_like(c)
    return torch.stack([
        torch.stack([c, -s, z]),
        torch.stack([s,  c, z]),
        torch.stack([z,  z, o]),
    ])
```
